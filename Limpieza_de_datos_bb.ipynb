```python
import pandas as pd
```

```python
df = pd.read_csv("bike_buyers.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
# Cambiamos de tipo objeto a float el income
df["Income"] = pd.to_numeric(df["Income"].str.replace(",", ""))
```

```python
df.describe()
```

```python
# Filtra las filas duplicadas
duplicates = df[df.duplicated(subset=['ID'], keep=False)]

# Ordena las filas duplicadas por el ID
sorted_duplicates = duplicates.sort_values(by='ID')

# Muestra el DataFrame con los duplicados ordenados
sorted_duplicates
```

```python
# Guardamos un nuevo archivo con los duplicados para futuras inspecciones
# antes de eliminarlos
sorted_duplicates.to_csv("bike_buyers_duplicateds.csv", index=False)
# Incluye los que deberiamos quedarnos tambien
```

```python
# Para eliminar columnas
df = df.drop_duplicates() # Coinciden todos los valores
#df.drop_duplicates(subset=["ID"]) # Valores de movies que se repiten
```
